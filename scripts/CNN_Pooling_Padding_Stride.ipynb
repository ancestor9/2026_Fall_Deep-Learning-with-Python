# Padding과 Stride의 관계

 Convolution Layer나 Pooling Layer나 `padding`, `stride` 옵션 작동 원리 자체는 동일하다. 옵션 적용 순서는 `padding` -> `stride`이다.

 이 내용을 이해하는 데 한해, 입력은 convolution 혹은 pooling을 하기 전의 데이터(주로 이미지), 중간 출력은 stride 옵션을 적용해 convolution 혹은 pooling을 하기 전, 패딩만 적용한 데이터라고 정의한다. 출력은 padding, stride 옵션을 거쳐 convolution, pooling 연산을 모두 적용한 후의 결과라고 정의한다.

 연산 적용 전후의 이미지 사이즈 변화는 이전의 강의에서도 보았듯, `{(N+2P-F)/stride}+1`이다.


1. Padding
    - SAME: 입력과 중간 출력의 사이즈가 같아지도록 패딩을 붙인다. 인터넷을 찾아 보면 일단 2배로 이미지를 확장한다고 되어 있는데, 정확히 말하면 공식 상에서 입력 이미지가 패딩을 거쳐 `{(N+2P-F)}+1` 중간 출력으로 갈 때, 중간 출력의 이미지와 입력 이미지의 각 변의 사이즈가 같아지도록 `P`를 찾아 적용하는 게 아닐까 싶다. (documentation, 소스코드 찾아 보고 이해할 필요가 있다.)
    - VALID: 그런 거 신경 안 쓰고, 그냥 `{(N-F)}+1` 공식에 의해 중간 출력 결과를 내 놓는다.


2. Stride

 패딩된 결과에 따라 갈 수 있는 한도 안에서 stride를 적용하며 convolution 혹은 pooling 연산을 한다. 그렇게 적용하고 맞지 않는 사이즈의 data는 drop(혹은 crop)한다.

```
import numpy as np
import sys
from tensorflow.keras.layers import Conv2D, MaxPooling2D
import tensorflow as tf
```

```
np.set_printoptions(threshold=sys.maxsize)
```

```
k_size = 4
p_size = 2
sample_data = tf.constant(np.arange(1.0, 201.0).reshape(-1, 20, 10, 1), dtype=tf.float32)
```

## Convolution : Padding과 Stride의 관계

```
# conv padding same
sample_conv = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=1, padding='same', activation='linear')(sample_data)
sample_conv2 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=2, padding='same', activation='linear')(sample_data)
sample_conv3 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=3, padding='same', activation='linear')(sample_data)
sample_conv4 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=4, padding='same', activation='linear')(sample_data)
sample_conv5 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=5, padding='same', activation='linear')(sample_data)
sample_conv6 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=6, padding='same', activation='linear')(sample_data)
sample_conv7 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=7, padding='same', activation='linear')(sample_data)
sample_conv8 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=8, padding='same', activation='linear')(sample_data)
sample_conv9 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=9, padding='same', activation='linear')(sample_data)
sample_conv10 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=10, padding='same', activation='linear')(sample_data)
sample_conv11 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=11, padding='same', activation='linear')(sample_data)
sample_conv19 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=19, padding='same', activation='linear')(sample_data)
sample_conv20 = Conv2D(filters=1, kernel_size=(k_size, k_size), strides=20, padding='same', activation='linear')(sample_data)
print(sample_conv.shape) # 패딩이 same으로 잡히니까, 출력 이미지 사이즈가 원본과 동일하다: (20, 10)
print(sample_conv2.shape) # 동일한 출력 이미지 사이즈 (20, 10)에서 strides가 2로 적용되니까 : (10, 5)
print(sample_conv3.shape) # 동일한 출력 이미지 사이즈 (20, 10)에서 strides가 3으로 적용되니까 가능한 데까지만 나온다 : 0, 3, 6, 9, 12, 15, 18 // 0, 3, 6, 9 // (7, 4)
print(sample_conv4.shape) # 동일한 출력 이미지 사이즈 (20, 10)에서 strides가 4로 적용되니가 0, 4, 8, 12, 16 // 0, 4, 8 // (5, 3)
print(sample_conv5.shape) # (20, 10) -> strides 5 적용 -> 0, 5, 10, 15 // 0, 5 -> (4, 2)
print(sample_conv6.shape) # (20, 10) -> strides 6 적용 -> 0, 6, 12, 18 // 0, 6 -> (4, 2)
print(sample_conv7.shape) # (20, 10) -> strides 7 적용 -> 0, 7, 14 // 0, 7 -> (3, 2)
print(sample_conv8.shape)
print(sample_conv9.shape)
print(sample_conv10.shape)
print(sample_conv11.shape)
print('...')
print(sample_conv19.shape)
print(sample_conv20.shape)
```

```
import tensorflow as tf
from tensorflow.keras import layers, models

# ==========================================
# 1. Standard CNN (Conv + MaxPool + Dense Layer)
# ==========================================
def create_standard_cnn(num_classes=10):
    model = models.Sequential([
        # 입력: [28, 28, 1]

        # Conv Block 1
        layers.Conv2D(filters=16, kernel_size=(3, 3), strides=(1, 1), padding='same',
                      activation='relu', input_shape=(28, 28, 1)), # -> [28, 28, 16]
        layers.BatchNormalization(),
        layers.MaxPooling2D(pool_size=(2, 2), strides=(2, 2)),      # -> [14, 14, 16]

        # Conv Block 2
        layers.Conv2D(filters=32, kernel_size=(3, 3), strides=(1, 1), padding='same',
                      activation='relu'),                          # -> [14, 14, 32]
        layers.BatchNormalization(),
        layers.MaxPooling2D(pool_size=(2, 2), strides=(2, 2)),      # -> [7, 7, 32]

        # Dense Layer
        layers.Flatten(),                                           # -> [1568] (32 * 7 * 7)
        layers.Dense(128, activation='relu'),                       # -> [128]
        layers.Dense(num_classes, activation='softmax')             # -> [10]
    ], name="Standard_CNN")

    return model


# ==========================================
# 2. Fully Convolutional Network (FCN)
# Dense Layer를 1x1 / Global Conv로 대체
# ==========================================
def create_fcn(num_classes=10):
    model = models.Sequential([
        # 입력: [None, None, 1] - 이미지 크기에 제약이 없도록 None 설정 가능

        # Conv Block 1
        layers.Conv2D(16, (3, 3), strides=1, padding='same', activation='relu', input_shape=(28, 28, 1)), # -> [28, 28, 16]
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2), strides=2),                                                            # -> [14, 14, 16]

        # Conv Block 2
        layers.Conv2D(32, (3, 3), strides=1, padding='same', activation='relu'),                           # -> [14, 14, 32]
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2), strides=2),                                                            # -> [7, 7, 32]

        # Fully Convolutional Layer 1: Dense(128)을 대체
        # (7, 7) 공간 크기를 1x1로 압축
        layers.Conv2D(128, kernel_size=(7, 7), strides=1, padding='valid', activation='relu'),            # -> [1, 1, 128]

        # Fully Convolutional Layer 2: Dense(10)을 대체
        layers.Conv2D(num_classes, kernel_size=(1, 1), strides=1, padding='valid'),                       # -> [1, 1, 10]

        # 분류 출력을 위해 차원 정리 [1, 1, 10] -> [10]
        layers.Reshape((num_classes,)),
        layers.Activation('softmax')
    ], name="Fully_Convolutional_Network")

    return model


# ==========================================
# 차원 변화 및 구조 확인
# ==========================================
if __name__ == "__main__":
    # 1. Standard CNN 요약 출력
    print("=" * 60)
    std_cnn = create_standard_cnn()
    std_cnn.summary()

    # 2. FCN 요약 출력
    print("\n" + "=" * 60)
    fcn = create_fcn()
    fcn.summary()

    # 3. Dummy 데이터 입력 테스트
    dummy_input = tf.random.normal([4, 28, 28, 1]) # Batch, Height, Width, Channels

    std_out = std_cnn(dummy_input)
    fcn_out = fcn(dummy_input)

    print("\n--- Output Tensor Shapes ---")
    print(f"Standard CNN Output Shape : {std_out.shape}")
    print(f"FCN Output Shape          : {fcn_out.shape}")
```

```
from tensorflow.keras.layers import Conv2D, MaxPooling2D
import tensorflow as tf
import numpy as np

# 주석의 (Batch, 28, 28, 1) 크기에 맞추어 sample_data 재정의
sample_data = tf.constant(np.arange(1.0, 785.0).reshape(1, 28, 28, 1), dtype=tf.float32)

# 1. Standard Conv & Pooling 연산 흐름 (Stride=1)
# sample_data: (Batch, 28, 28, 1)
sample_conv = Conv2D(filters=32, kernel_size=(3, 3), strides=1, padding='same', activation='relu')(sample_data)
sample_pool = MaxPooling2D(pool_size=(2, 2), strides=2, padding='same')(sample_conv)

# 2. Stride 변경에 따른 축소 과정
sample_conv2 = Conv2D(filters=64, kernel_size=(3, 3), strides=1, padding='same', activation='relu')(sample_pool)
sample_pool2 = MaxPooling2D(pool_size=(2, 2), strides=2, padding='same')(sample_conv2)

# 3. Fully Convolutional Network (FCN) 변환: Dense 레이어를 1x1 Conv로 대체
# (7, 7) 공간을 1x1로 압축하여 채널을 확장/축소
sample_fcn1 = Conv2D(filters=128, kernel_size=(7, 7), strides=1, padding='valid', activation='relu')(sample_pool2)
sample_fcn2 = Conv2D(filters=10, kernel_size=(1, 1), strides=1, padding='valid', activation='softmax')(sample_fcn1)

# Shape 확인
print("Original Input :", sample_data.shape) # (None, 28, 28, 1)
print("Conv1 (3x3, s1):", sample_conv.shape) # (None, 28, 28, 32)
print("Pool1 (2x2, s2):", sample_pool.shape) # (None, 14, 14, 32)  <-- H, W 절반 감소
print("Conv2 (3x3, s1):", sample_conv2.shape) # (None, 14, 14, 64) <-- 채널수 64 확장
print("Pool2 (2x2, s2):", sample_pool2.shape) # (None, 7, 7, 64)   <-- H, W 절반 감소
print("FCN1  (7x7, s1):", sample_fcn1.shape) # (None, 1, 1, 128)  <-- 공간 크기 (1, 1)로 압축
print("FCN2  (1x1, s1):", sample_fcn2.shape) # (None, 1, 1, 10)   <-- 클래스 수(10개)로 최종 출력
```
